```python
# --- CÉLULA DE CONVERSÃO (EXECUTAR SÓ UMA VEZ) ---

import json
import pandas as pd
from pandas import json_normalize

try:
    # 1. Carrega o JSON bruto
    with open('base_de_casos.json', 'r', encoding='utf-8') as f:
        dados_brutos = json.load(f)

    # 2. "Achata" a estrutura aninhada
    df = json_normalize(dados_brutos)

    # 3. Limpa os nomes das colunas
    df.columns = df.columns.str.replace('problema.', '', regex=False)
    df.columns = df.columns.str.replace('solucao.', '', regex=False)

    # 4. Define o id_caso como o índice
    df = df.set_index('id_caso')
    
    # 5. SALVA O CSV
    df.to_csv('base_de_casos.csv')
    
    print("Sucesso! O arquivo 'base_de_casos.csv' foi criado.")
    display(df.head())

except FileNotFoundError:
    print("Erro: 'base_de_casos.json' não encontrado.")
except Exception as e:
    print(f"Um erro ocorreu: {e}")
```

## Configuração: Imports e Constantes

Nesta célula, importamos as bibliotecas (Pandas, Math, JSON) e definimos as constantes globais do nosso sistema:

1.  **`NOME_ARQUIVO_BASE_CASOS`**: O arquivo CSV que armazena nossa Base de Casos (Case Library).
2.  **`PESOS`**: Define a importância (peso) de cada atributo na hora de calcular a similaridade. A soma total deve ser 1.0.
3.  **`TABELA_SIMILARIDADE_BAIRRO`**: Implementa a **Similaridade Local** para atributos simbólicos não ordenados. Definimos manualmente o quão similar um bairro é do outro (ex: Centro é 70% similar ao Jardim Carvalho, mas só 20% similar a Uvaranas).

```python
import json
import math
import pandas as pd
# Não precisamos mais de json_normalize aqui

# --- Constantes ---

# ATUALIZADO: Agora vamos usar o arquivo CSV
NOME_ARQUIVO_BASE_CASOS = "base_de_casos.csv"

# Pesos de cada atributo para o cálculo de similaridade
PESOS = {
    "metros_quadrados": 0.30,
    "bairro": 0.20,
    "mobiliado": 0.15,
    "num_quartos": 0.15,
    "num_banheiros": 0.10,
    "vagas_garagem": 0.10
}

# Garante que a soma dos pesos é 1.0 (ou próximo disso)
soma_pesos = sum(PESOS.values())
if not math.isclose(soma_pesos, 1.0):
    print(f"Atenção: A soma dos pesos é {soma_pesos}, mas deveria ser 1.0. Ajustando...")
    for chave in PESOS:
        PESOS[chave] /= soma_pesos


# Tabela de Similaridade para o atributo simbólico "bairro"
TABELA_SIMILARIDADE_BAIRRO = {
    "Centro": {
        "Centro": 1.0,
        "Jardim Carvalho": 0.7,
        "Oficinas": 0.3,
        "Uvaranas": 0.2
    },
    "Jardim Carvalho": {
        "Centro": 0.7,
        "Jardim Carvalho": 1.0,
        "Oficinas": 0.4,
        "Uvaranas": 0.3
    },
    "Oficinas": {
        "Centro": 0.3,
        "Jardim Carvalho": 0.4,
        "Oficinas": 1.0,
        "Uvaranas": 0.8
    },
    "Uvaranas": {
        "Centro": 0.2,
        "Jardim Carvalho": 0.3,
        "Oficinas": 0.8,
        "Uvaranas": 1.0
    }
}

print("Imports e Constantes carregados (com Pandas e CSV).")
```

## Carregando a Base de Casos e Pré-processamento

Aqui, definimos duas funções:

1.  **`carregar_base_de_casos`**: Lê o arquivo CSV e o carrega em um DataFrame do Pandas.
2.  **`calcular_ranges_normalizacao`**: Esta função é crucial. Ela encontra os valores `min` e `max` de cada atributo numérico na base. [cite_start]Isso é necessário para a **Normalização** [cite: 254-257]. [cite_start]Como a aula 08 menciona, "Atributos com escala mais ampla/maior variabilidade tendem a ter maior peso nos cálculos de distâncias"[cite: 256]. [cite_start]Usaremos a "Reescala Linear [0,1]" [cite: 257] para evitar isso.

```python
def carregar_base_de_casos(arquivo_csv):
    """
    Carrega a base de casos diretamente de um arquivo CSV para um DataFrame.
    """
    try:
        # Lê o CSV, já definindo a coluna 'id_caso' como nosso índice
        df = pd.read_csv(arquivo_csv, index_col='id_caso')
        
        print(f"Base de casos (CSV) carregada em um DataFrame. {len(df)} casos encontrados.")
        return df
    except FileNotFoundError:
        print(f"Erro: Arquivo '{arquivo_csv}' não encontrado. Execute a célula de conversão primeiro.")
        return pd.DataFrame() # Retorna um DataFrame vazio em caso de erro

def calcular_ranges_normalizacao(df):
    """
    Calcula os valores Mínimo e Máximo para cada atributo numérico
    diretamente do DataFrame.
    """
    ranges = {}
    atributos_numericos = ["metros_quadrados", "num_quartos", "num_banheiros", "vagas_garagem"]
    
    if df.empty:
        return None

    for atributo in atributos_numericos:
        # MANTEMOS OS TIPOS ORIGINAIS DO PANDAS (int64/float64)
        ranges[atributo] = {
            "min": df[atributo].min(),
            "max": df[atributo].max()
        }
    
    # Em vez de json.dumps, usamos o display() do Pandas para formatar
    print("Valores Min-Max para normalização calculados:")
    
    # Converte o dicionário de ranges em um DataFrame para exibição bonita
    ranges_df = pd.DataFrame(ranges).transpose() 
    display(ranges_df)
    
    return ranges # Retornamos o dicionário com os tipos int64/float64

# --- Execução inicial ---
df_base_de_casos = carregar_base_de_casos(NOME_ARQUIVO_BASE_CASOS)
ranges_normalizacao = calcular_ranges_normalizacao(df_base_de_casos)

# Mostra as 5 primeiras linhas do DataFrame para verificar
if not df_base_de_casos.empty:
    display(df_base_de_casos.head())
```

## Etapa 1: RECUPERAÇÃO (Retrieve)

[cite_start]Esta é a primeira etapa do ciclo RBC[cite: 19, 258, 1101]. O objetivo é encontrar na base de casos o problema anterior mais similar ao novo problema.

1.  [cite_start]**`normalizar_valor`**: Aplica a fórmula de Reescala Linear [0,1] (`(valor - min) / (max - min)`)[cite: 257].
2.  [cite_start]**`calcular_similaridade_local`**: Calcula a similaridade para um *único* atributo[cite: 175].
    * [cite_start]Para **Números** (ex: `metros_quadrados`): `1 - abs(valor_novo_norm - valor_caso_norm)`[cite: 189].
    * [cite_start]Para **Símbolos** (ex: `bairro`, `mobiliado`): Usa a tabela de similaridade ou a correspondência exata[cite: 208, 248].
3.  [cite_start]**`calcular_similaridade_global`**: Combina todas as similaridades locais usando os `PESOS` que definimos [cite: 172-174].
4.  [cite_start]**`recuperar_caso_mais_similar`**: Implementa a **Recuperação Sequencial**[cite: 318, 337]. [cite_start]Ela aplica a `calcular_similaridade_global` a todos os casos da base e retorna o "vizinho mais próximo" (Nearest Neighbour)[cite: 82, 358].

```python
def normalizar_valor(valor, ranges_atributo):
    """
    Normaliza um valor único usando a fórmula min-max: (valor - min) / (max - min)
    """
    min_val = ranges_atributo["min"]
    max_val = ranges_atributo["max"]
    
    # Evita divisão por zero se todos os valores na base forem iguais
    if max_val == min_val:
        return 0
    
    return (valor - min_val) / (max_val - min_val)

def calcular_similaridade_local(valor_novo, valor_caso, atributo, ranges_normalizacao):
    """
    Calcula a similaridade local para um atributo específico.
    """
    # Atributos numéricos
    if atributo in ranges_normalizacao:
        ranges_attr = ranges_normalizacao[atributo]
        
        v_novo_norm = normalizar_valor(valor_novo, ranges_attr)
        v_caso_norm = normalizar_valor(valor_caso, ranges_attr)
        
        return 1 - abs(v_novo_norm - v_caso_norm)
    
    # Atributo simbólico (bairro)
    elif atributo == "bairro":
        try:
            return TABELA_SIMILARIDADE_BAIRRO[valor_novo][valor_caso]
        except KeyError:
            return 0.0
            
    # Atributo simbólico (mobiliado)
    elif atributo == "mobiliado":
        return 1.0 if valor_novo == valor_caso else 0.0
    
    else:
        return 0.0

def calcular_similaridade_global(linha_caso, problema_novo, pesos, ranges_normalizacao):
    """
    Calcula a similaridade global. Esta função será aplicada a cada LINHA do DataFrame.
    'linha_caso' é uma Series do Pandas.
    'problema_novo' é um dicionário.
    """
    similaridade_total = 0.0
    
    for atributo, peso in pesos.items():
        if atributo in problema_novo and atributo in linha_caso:
            sim_local = calcular_similaridade_local(
                problema_novo[atributo],
                linha_caso[atributo],
                atributo,
                ranges_normalizacao
            )
            similaridade_total += sim_local * peso
            
    return similaridade_total

def recuperar_caso_mais_similar(df, problema_novo, pesos, ranges_normalizacao):
    """
    Implementa a Recuperação usando Pandas .apply().
    """
    if df.empty:
        return None, 0.0

    # 1. Cria uma cópia para não alterar o DataFrame original
    df_temp = df.copy()
    
    # 2. Aplica a função de similaridade a TODAS as linhas (axis=1)
    df_temp['similaridade'] = df_temp.apply(
        calcular_similaridade_global,
        axis=1,
        # Argumentos extras para a função:
        problema_novo=problema_novo,
        pesos=pesos,
        ranges_normalizacao=ranges_normalizacao
    )
    
    # 3. Encontra o ID (índice) do caso com a maior similaridade
    id_melhor_caso = df_temp['similaridade'].idxmax()
    
    # 4. Obtém a similaridade máxima
    max_similaridade = df_temp['similaridade'].max()
    
    # 5. Obtém o melhor caso como um dicionário
    melhor_caso_dict = df_temp.loc[id_melhor_caso].to_dict()
    
    # Adiciona o ID ao dicionário para exibição
    melhor_caso_dict['id_caso'] = id_melhor_caso
            
    return melhor_caso_dict, max_similaridade, df_temp

print("Funções da Etapa 1 (Recuperação) definidas.")
```

## Função Auxiliar: Obter Entrada do Usuário

[cite_start]Esta célula define uma função simples para coletar os dados do **Novo Problema** (ou `New Case` [cite: 17, 262]) do usuário.

```python
def obter_novo_problema():
    """
    Solicita ao usuário os dados do novo problema (imóvel).
    """
    print("\n--- Por favor, insira os dados do novo imóvel ---")
    
    try:
        problema = {}
        problema["metros_quadrados"] = float(input("Metros quadrados (ex: 70): "))
        problema["num_quartos"] = int(input("Número de quartos (ex: 3): "))
        problema["num_banheiros"] = int(input("Número de banheiros (ex: 1): "))
        problema["vagas_garagem"] = int(input("Vagas de garagem (ex: 1): "))
        
        bairros_validos = list(TABELA_SIMILARIDADE_BAIRRO.keys())
        while True:
            bairro = input(f"Bairro {bairros_validos}: ")
            if bairro in bairros_validos:
                problema["bairro"] = bairro
                break
            else:
                print(f"Bairro inválido. Por favor, escolha um dos seguintes: {bairros_validos}")
        
        while True:
            mobiliado_input = input("É mobiliado? (s/n): ").lower()
            if mobiliado_input == 's':
                problema["mobiliado"] = True
                break
            elif mobiliado_input == 'n':
                problema["mobiliado"] = False
                break
            else:
                print("Entrada inválida. Digite 's' para sim ou 'n' para não.")
                
        return problema
    
    except ValueError:
        print("Erro: Entrada inválida. Por favor, insira números corretos.")
        return None

print("Função 'obter_novo_problema' definida.")
```

## Etapa 2: REUSO (Reuse)

[cite_start]Esta é a segunda etapa do ciclo[cite: 29, 446, 467]. O objetivo é pegar a solução do caso recuperado e adaptá-la para o novo problema.

[cite_start]O requisito do trabalho é claro: "Implementar uma forma de **adaptação automática (exceto a adaptação nula)**"[cite: 410]. [cite_start]A "Adaptação Nula" [cite: 533-535] seria simplesmente copiar o preço antigo, o que não é permitido.

[cite_start]Portanto, implementamos uma **Adaptação Transformacional Substitucional** [cite: 557-561][cite_start], que usa "conhecimento do domínio... na forma de regras ou fórmulas"[cite: 553].

Nossa regra de adaptação é:
1.  **Substituição Base:** Calcular o `preço/m²` do caso antigo e aplicar ao `m²` do caso novo. [cite_start](Similar ao exemplo do pacote de viagem [cite: 567-569]).
2.  [cite_start]**Regra de Ajuste:** Aplicar um fator de ajuste (ex: +/- 10%) se o atributo `mobiliado` for diferente entre o caso antigo e o novo [cite: 618-622].

```python
# --- ETAPA 2: REUSO (Adaptação) ---

# Define um fator de ajuste de 10% para o atributo 'mobiliado'
FATOR_AJUSTE_MOBILIADO = 0.10 # 10%

def reusar_solucao_adaptada(caso_recuperado, novo_problema):
    """
    Adapta a solução do caso recuperado para o novo problema.
    Isto é uma "Adaptação Transformacional Substitucional" baseada em regras.
    """
    
    # 1. Pega os dados do caso antigo
    preco_antigo = caso_recuperado['preco_vendido']
    mq_antigo = caso_recuperado['metros_quadrados']
    mobiliado_antigo = caso_recuperado['mobiliado']
    
    # 2. Pega os dados do novo problema
    mq_novo = novo_problema['metros_quadrados']
    mobiliado_novo = novo_problema['mobiliado']
    
    # Evita divisão por zero (caso um imóvel tenha 0 m²)
    if mq_antigo == 0:
        return preco_antigo 

    # 3. Calcula o preço/m² do caso antigo
    preco_por_mq_antigo = preco_antigo / mq_antigo
    
    # 4. Adaptação base (Substituição): Calcula o novo preço com base no novo m²
    preco_base_adaptado = preco_por_mq_antigo * mq_novo
    
    # 5. Regra de Adaptação 2: Ajusta o preço com base no atributo 'mobiliado'
    
    if mobiliado_novo and not mobiliado_antigo:
        # O novo é mobiliado e o antigo não era -> Adiciona 10%
        preco_final_adaptado = preco_base_adaptado * (1 + FATOR_AJUSTE_MOBILIADO)
        print(f"INFO: Aplicado ajuste de +{FATOR_AJUSTE_MOBILIADO*100}% (Mobiliado)")
        
    elif not mobiliado_novo and mobiliado_antigo:
        # O novo NÃO é mobiliado e o antigo era -> Subtrai 10%
        preco_final_adaptado = preco_base_adaptado * (1 - FATOR_AJUSTE_MOBILIADO)
        print(f"INFO: Aplicado ajuste de -{FATOR_AJUSTE_MOBILIADO*100}% (Não Mobiliado)")
        
    else:
        # Ambos são iguais (mobiliados ou não-mobiliados), preço base se mantém
        preco_final_adaptado = preco_base_adaptado
        print("INFO: Nenhum ajuste de mobília necessário.")

    return preco_final_adaptado

print("Funções da Etapa 2 (Reuso) definidas.")
```

## LOOP PRINCIPAL (Etapas 1 + 2)

Esta é a célula principal que executa o fluxo do sistema. Você pode executar esta célula várias vezes para testar diferentes entradas.

O fluxo é:
1.  Chama `obter_novo_problema()` (Célula 10).
2.  Passa o problema para `recuperar_caso_mais_similar()` (Célula 8).
3.  Pega o caso recuperado e o novo problema e passa para `reusar_solucao_adaptada()` (Célula 12).
4.  [cite_start]Exibe a **Solução Sugerida** (`Suggested Solution` [cite: 37, 1119]).

```python
# --- Loop Principal (Integrando Etapas 1 e 2) ---

# Você pode re-executar esta célula sempre que quiser testar com um novo imóvel

# Variáveis globais para armazenar o estado
novo_problema_global = None
caso_recuperado_global = None
solucao_sugerida_global = None

if not df_base_de_casos.empty:
    
    # 1. Obter o novo problema do usuário (da Célula 10)
    novo_problema_global = obter_novo_problema()
    
    if novo_problema_global:
        print("\n--- ETAPA 1: RECUPERAÇÃO (Retrieve) ---")
        
        # 2. Recuperar o caso mais similar (da Célula 8)
        caso_recuperado_global, similaridade, df_com_similaridade = recuperar_caso_mais_similar(
            df_base_de_casos,
            novo_problema_global,
            PESOS,
            ranges_normalizacao
        )
        
        if caso_recuperado_global:
            print(f"\nProblema Novo: {novo_problema_global}")
            print(f"\nCaso mais similar encontrado (ID: {caso_recuperado_global['id_caso']}) com similaridade de {similaridade*100:.2f}%:")
            
            # Formata a exibição do caso recuperado
            print("  -> Problema:")
            for attr in PESOS.keys():
                print(f"     {attr}: {caso_recuperado_global[attr]}")
            print(f"  -> Solução Antiga: R$ {caso_recuperado_global['preco_vendido']:,.2f}")
            
            # Mostra o DataFrame com a nova coluna 'similaridade'
            print("\n--- Detalhes do Cálculo de Similaridade ---")
            display(df_com_similaridade.sort_values(by='similaridade', ascending=False).head(5))
            
            # --- Início da Nova Etapa ---
            print("\n--- ETAPA 2: REUSO (Reuse/Adapt) ---")
            
            # 3. Reusar e adaptar a solução (da Célula 12)
            solucao_sugerida_global = reusar_solucao_adaptada(caso_recuperado_global, novo_problema_global)
            
            print(f"\nPreço Sugerido (Adaptado): R$ {solucao_sugerida_global:,.2f}")
            
        else:
            print("Nenhum caso similar encontrado.")
else:
    print("O programa não pode iniciar sem uma base de casos.")
```
